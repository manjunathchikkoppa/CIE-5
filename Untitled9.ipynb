```python
import spacy
from   nltk import ngrams
nlp = spacy.load("en_core_web_sm")
doc = nlp("Ai is simulation of human intelligence by machine. It includes expects system NLP speech recoynition and machine vision")
bigrams = ngrams(doc,2)
print(list(bigrams))
```
